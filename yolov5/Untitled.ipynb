```python
import torch
import cv2
import numpy as np
```

```python
model = torch.hub.load('ultralytics/yolov5', 'custom', path='weights/best.pt')
```

```python
image = "/home/connor/yolov5/Database1/temp1/train/images/17.JPEG"
# Object Detection

results = model(image)
```

```python
img = np.copy(results.render()[0])
```

```python
for bbox in results.xyxy[0].cpu().numpy():
            x_min, y_min, x_max, y_max, conf, cls = bbox
            # Calculate the centroid
            centroid_x = int((x_min + x_max) / 2)
            centroid_y = int((y_min + y_max) / 2)
            # Draw the centroid
            cv2.circle(img, (centroid_x, centroid_y), 5, (0, 255, 0), 1)
```

```python
cv2.imshow('Detected Frame', img)  
```
